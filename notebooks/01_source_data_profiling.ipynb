```python
import pandas as pd
import glob
import os

# Folder containing Olist CSV files
folder_path = r"F:\enterprise-retail-data-warehouse\data\raw\olist"

# Get all CSV files
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

print(f"Total CSV files found: {len(csv_files)}")

for file in csv_files:

    print("\n" + "=" * 80)
    print(f"FILE: {os.path.basename(file)}")
    print("=" * 80)

    df = pd.read_csv(file)

    print(f"Rows       : {df.shape[0]}")
    print(f"Columns    : {df.shape[1]}")

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nNULL Values:")
    print(df.isnull().sum())

    print(f"\nDuplicate Rows: {df.duplicated().sum()}")
```
